```python
import random
number=random.randint(0,101)
while True:

  user_number=int(input("Enter a number you thought : "))
  if user_number==number:
    print("Congratulations! you guess the right number.")
    break
  elif user_number>number:
    print("Sorry, number is too high.")
  else:
    print("Sorry, number is too low.")

print("Game over !")
```
