**Numpy**is a numerical Python. we can use it to store all sorts of data in one dimensional array, two dimensional array and n-dimensional arrays. we use numpy instead of list because numpy take less space which is easy to read fast. and it takes contigious space and numpy store same datatype so we dont need to check the data type of each elements when we are iterating through the iteraor.

```python
#initalize of an array
import numpy as np
a=np.array([2,4,6,8])#one dimensional aray
print(a)
```

```python
#get size
a.size
```

```python
#get shape
a.shape
```

```python
#creation of 2D array
a=np.array([[2,4,6,8], [4,3,28,4]])
a.dtype
a.itemsize
```

Accessing a specific elemts with the help of Slicing / indexing:
In numpy we can get a specific elements and specific columns and rows with the help of slicing/indexing.
slicing of aray : [rows, column]

```python
a=np.array([[[2,4,6,8], [10,5,7,9],[13,16,17,18],[44,55,77,88]]])
#if i want to get 17
print(a[0,2,2])
#for row and columns
print(a[0,:,3]) #for last column
print(a[0,1,:]) # returns second row
```

**usage of numpy in mathematics**: we can do all kinds of arithmatics operations with the help of numpy. we can do with two or different arays together or we can add manually a differnt numbers in our array

```python
a=np.array([2,4,6])
print(a+2)#added 2 in every elements
```

```python
#for multiple arrays:
a=np.array([2,4,6])
b=np.array([4,6,7])
print(a*b)
```

```python
#multiplication of the matrixes
a=np.array([[4,5,6,8],[8,9,5,6]])
b=np.array([[8,6,7,5],[9,8,9,5]])
b=b.T #Transpose of matrix (means changage of rows and columns.)
print(np.linalg.matmul(a,b))
```

**Random Number Generation**

```python
a=np.random.rand(2,3)
print(a) # returns the random float int between 0 and 1
```

```python
a=np.random.randint(20,size=(2,3,4)) #we pass two parameter one is the range and another is the size of our array
print(a)
```

```python
a=np.arange(20)#returns the array with the int between 0 to 19
print(a)
```

```python
a=np.linspace(0,20,5) #create evenly spaced numbers between two values.
print(a)
```

**Concatenate of 2 arrays**

```python
a=np.array([2,3,5,6,7])
b=np.array([4,8,9,10,11])
np.concatenate((a,b))
```

```python
np.std((a,b))#returns standard deviation
```

**Reshaping of an array** : we can change the shape of our array with the help of .reshape function . for that we need to have the same numbers of elements of the new array and our original aray.

```python
a=np.array([98,99,34,56,78,45,67,78])
a.reshape(2,2,2)#This is valid for reshaping
a.reshape(3,2)#This is invalid because the numbers of elemnts will be differents here
```

**Linear Algebra Basis With Numpy** : we can solve multiple linear algerbra problems with the help of numpy like inverse of the matrix, determinant of the matrix, multiplication fo the matrix.

```python
a=np.identity(4)
np.linalg.det(a) #return the determinent of the matrix
```

```python
a=np.array([[4,5],[7,8]])
np.linalg.inv(a) #returns the inverse of the matrix
```

**Some Important Functions**

```python
#sum
a=np.array([[2,4,6],[6,7,8],[4,32,1]])
np.sum(a)#returns the total sum of an array
np.sum(a, axis=0)#returns the sum of the column
np.sum(a, axis=1)#returns the sum of the rows
```

```python
#min
np.min(a)
np.max(a)
np.argmax(a)#returns the index of the maximum element
np.argmin(a)#returns the index of the smallest element
```
